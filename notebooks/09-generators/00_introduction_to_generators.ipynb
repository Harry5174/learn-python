## Python Generators
- Generators use yeild to suspend and resume their execution and state around last point of value

```python
def create_cubes(n):
    # result = []
    
    for x in range(n):
        yield x**3
    
    # return result
```

```python
for x in create_cubes(10):
    print(x)
```

```python
# create_cubes(10)
list(create_cubes(10))
```

```python
def fibon(n):
    a = 1
    b = 1
    
    for i in range(n):
        yield a
        a, b = b, a+b
```

```python
for number in fibon(10):
    print(number)
```

```python
def simple_gen():
    for x in range(3):
        yield x
```

```python
for num in simple_gen():
    print(num)
```

```python
g = simple_gen()
```
