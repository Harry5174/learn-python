## Color Transparency

### RGBA - Red, Green, Blue, Aplha

```python
red = Image.open('red_color.jpg')
```

```python
red
```

```python
blue = Image.open('blue_color.png')
```

```python
blue
```

```python
blue
```

```python
blue.putalpha(225)
red.putalpha(225)
```

```python
blue.show()
```
