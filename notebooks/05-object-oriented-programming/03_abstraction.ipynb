# Abstraction

```python
class Animal():
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        raise NotImplementedError("Subclass must implement this abstract method")
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):  # abstract class

    @abstractmethod
    def speak(self):
        pass
```

```python
class Dog(Animal):
    
    def speak(self):
        return "Bark"
```

```python
d = Dog()
print(d.speak())
```

```python
class Cat(Animal):
    pass

c = Cat() # will raise error
```

```python
# Another Examle

class STT(ABC):

    @abstractmethod
    def transcribe(self, audio):
        pass
```

```python
class WhisperSTT(STT):
    def transcribe(self, audio):
        return "text"

class DeepgramSTT(STT):
    def transcribe(self, audio):
        return "text"
```

```python
# loose coupling
def process(stt: STT):
    return stt.transcribe(audio)
```
