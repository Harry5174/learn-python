# Method Types in Python Classes (or Class Method Types / Decorators in OOP)

## The Three Method Types

## @property → Attribute-like method

```python
class Account:
    def __init__(self):
        self.__balance = 100

    @property
    def balance(self):
        return self.__balance # called as obj.balance
```

```python
class Temperature:
    def __init__(self):
        self.__celsius = 0

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273:
            raise ValueError("Below absolute zero!")
        self.__celsius = value
```

## @classmethod → Works with CLASS, not object

```python
class Person:
    species = "Human"
    
    @classmethod
    def get_species(cls): # takes cls (class) not self (instance of the class)
        return cls.species
```

```python
Person.get_species()
```

```python
# real use case
class Person:
    def __init__(self, name):
        self.name = name
        
    @classmethod
    def from_string(cls, data):
        name = data.split("-")[0]
        return cls(name)
```

```python
p = Person.from_string("Harry-22")
```

```python
p.name
```

## @staticmethod → Utility function inside class

```python
class MathUtils:
    
    @staticmethod
    def add(a, b):
        return a+b
```

```python
MathUtils.add(2, 3)
```

# real world example

```python
class ModelConfig:
    
    def __init__(self, name, version):
        self.name = name
        self.version = version
        
    @classmethod
    def from_dict(cls, data):
        return cls(data['name'], data['version'])
    
    @staticmethod
    def validate_name(name):
        return len(name) > 0
    
    @property
    def full_name(self):
        return f"{self.name}:{self.version}"
```

```python
config = ModelConfig.from_dict({"name": "gpt", "version": "4"})

print(config.full_name)  # property
ModelConfig.validate_name("gpt")  # staticmethod
```
