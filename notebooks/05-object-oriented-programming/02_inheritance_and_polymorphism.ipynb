## Inheritance

```python
class Animal:
    def __init__(self):
        print('Animal created!')
        
    def who_am_i(self):
        print("I'm an animal!")
        
    def eat(self):
        print("I'm eating!")
```

```python
myanimal = Animal()
```

```python
myanimal.eat()
```

```python
# Inherit from parent class. Override super() class methods
class Dog(Animal):
    def __init__(self):
        super().__init__()
        print('Dog created!')
        
    def who_am_i(self):
        super().who_am_i()
        print("I'm a Dog!")
        
    def bark(self):
        print('Woof!')
        
```

```python
mydog = Dog()
```

```python
mydog.who_am_i()
```

```python
mydog.bark()
```

## Polymorphism

```python
class Dog:
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        return self.name + " says woof!"
```

```python
class Cat:
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        return self.name + " says meow!"
```

```python
felix = Dog('felix')
neko = Cat('neko')
```

```python
neko.speak()
```

```python
felix.speak()
```

```python
for pet in [neko, felix]:
    print(type(pet))
    print(pet.speak())
```

```python
def pet_speak(pet):
    return pet.speak()
```

```python
pet_speak(neko)
```

## Abstract Class

```python
class Animal():
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        raise NotImplementedError("Subclass must implement this abstract method")
```

```python
myanimal = Animal('Fred')
```

```python
myanimal.speak() # will raise error
```

```python
class Dog(Animal):
    def speak(self):
        return self.name + " says woof!"
```

```python
class Cat(Animal):
    def speak(self):
        return self.name + " says meow!"
```

```python
mydog = Dog("Tommy")
mycat = Cat("neko")
```

```python
mydog.speak()
```

```python
mydog.speak()
```

```python
pet_speak(mycat)
```
