## Special method

```python
class Book:
    def __init__(self, title, author, pages ):
        self.title = title
        self.author = author
        self.pages = pages
        
    def __str__(self):
        return f"{self.title} by {self.author}"
    
    def __len__(self):
        return self.pages
    
    def __del__(self):
        print("Book object has been deleted!")
```

```python
book = Book('Python rocks', 'Jose', 200)
```

```python
print(book)
```

```python
len(book)
```

```python
del book
```

```python
print(book)
```
