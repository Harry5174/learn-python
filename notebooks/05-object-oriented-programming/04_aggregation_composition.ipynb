# 1. Composition (Strong Relationship)

```python
class Engine:
    def start(self):
        return "Engine Started"
    
class Car:
    def __init__(self):
        self.engine = Engine() # created and instance inside when an instance of Car created
        
    def start(self):
        return self.engine.start()
    
# Engine is created inside Car
# Engine cannot exist independently (logically)
# lifetime is tied to car

# when Car dies -> Engine Dies
```

```python
class Car:
    def __init__(self):
        self.engine = Engine()  # tightly coupled
```

## 2. Aggregation (Weak Relationship)

```python
class Player:
    def __init__(self, name):
        self.name = name

class Team:
    def __init__(self, players):
        self.players = players  # passed from outside
```

```python
player = Player('Bruno'), Player('Felix')
team = Team(player)
```

```python
for player in team.players:
    print(player.name)
```

# Duck typing

```python
# Example 1 — Classical Polymorphism (Inheritance)

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Bark"

class Cat(Animal):
    def speak(self):
        return "Meow"
```

```python
def make_sound(animal: Animal):
    print(animal.speak())
    
# strict structure
# Enforced Interface
```

```python
# Example 2 — Duck Typing (no inheritance)

class Dog:
    def speak(self):
        return "Bark"

class Cat:
    def speak(self):
        return "Meow"
```

```python
def make_sound(animal):
    print(animal.speak())
    
# No base class
# No enforcement
# Still polymorphism works
```

```python
class Rock:
    pass

process_payment(Rock())  # will crash
```
