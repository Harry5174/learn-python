## Class Object Attributes

```python
class Dog:
    # SAME FOR ANY INSTANCE OF A CLASS
    species = 'mammal'
    
    
    def __init__(self, breed, name, spots):
        # Attributes 
        # We take in arguments
        # Assign it using self.attribute_name
        self.breed = breed
        self.name = name
        
        # Excepts boolean values
        self.spots = spots
        
    # OPERATIONS/Actions --> Methods
    def info(self):
        print(f'Name: {self.name}, Breed: {self.breed}, Spots: {self.spots}')
        
    def bark(self, number):
        print('Woof! {} times'.format(number))
        
dog = Dog(name='Alan', breed='Husky', spots=False)
dog.info()
```

```python
dog.bark(5)
```

## Public Attributes

```python
class Car:
    def __init__(self, speed):
        self.speed = speed
        
car = Car(120)
print(car.speed)
```

## Protected Attributes

```python
class Car:
    def __init__(self, speed):
        self._speed = speed  # This is internal please don't access this outside the class. Python does not prevent access (conceptual only)
        
car = Car(120)
print(car._speed) # python allows access
```

## Private attributes

```python
class Car:
    def __init__(self, speed):
        self.__speed = speed
        
car = Car(120)
```

```python
car.__speed # private attribute with double underscore will yield error
```
