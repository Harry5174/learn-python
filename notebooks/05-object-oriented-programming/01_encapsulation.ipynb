## Levels of access in python
## 1. Public (Default)

```python
class Account:
    def __init__(self):
        self.balance = 0  # public
```

```python
# Accessible from anywhere
acc = Account()
print(acc.balance)
```

## 2. Protected (Single underscore _ )

```python
class Account:
    def __init__(self):
        self._balance = 0 # protected (convention)
```

```python
# Meaning: don't touch outside class or subclass
# However python won't restrict
acc = Account()
print(acc._balance) # works but discouraged
```

## 3. Private (Double underscore __ )

```python
class Account:
    def __init__(self):
        self.__balance = 0 # private, internally it becomes _Account__balance
```

```python
acc = Account()
print(acc.__balance) # will give error
print(acc._Account__balance) # works. but not intended.
```

# Proper Encapsulation Pattern

```python
# instead of accessing variables directly. we use methods

class Account:
    def __init__(self):
        self.__balance = 0
        
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    
    def get_balance(self):
        return self.__balance
        
```

```python
acc = Account()
acc.deposit(100)
print(acc.get_balance())
```

## Pythonic method: properties

```python
class Account:
    def __init__(self):
        self.__balance = 0

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value >= 0:
            self.__balance = value
```

```python
acc = Account()
acc.balance = 100
print(acc.balance)
```

```python
class Temperature:
    def __init__(self):
        self.__celsius = 0

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273:
            raise ValueError("Below absolute zero!")
        self.__celsius = value
```

```python
temp = Temperature()
temp.celsius = 200
```

```python
temp.celsius
```

```python
## Task
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        # self._area = self.width*self.height
        return self.width * self.height
    
    @property
    def perimeter(self):
        return 2 * (self.width + self.height)
```

```python
rect = Rectangle(2,4)
rect.area
```
