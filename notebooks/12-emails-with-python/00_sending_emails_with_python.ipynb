## Sending emails

```python
# steps
# 1: connect to an email server
# 2: confirming connection
# 3: setting a protocol
# 4: logging on, and sending the messages
```

```python
import smtplib
```

```python
smtp_obj = smtplib.SMTP('smtp.gmail.com',587)
```

```python
smtp_obj.ehlo()
```

```python
smtp_obj.starttls()
```

```python
password = input('What is your password')
```

```python
import getpass
```

```python
password = getpass.getpass('Password Please: ')
```

```python
email = getpass.getpass('Email: ')
password = getpass.getpass('Password: ')

smtp_obj.login(email,password)
```

```python
from_address = email
to_address = email # to self

subject = input("Enter your line of subject: ")
message = input("Enter the body message: ")

msg = "Subject: "+ subject + '\n' + message

smtp_obj.sendmail(from_address, to_address, msg)
```

```python
smtp_obj.quit()
```
