## Viewing emails

```python
import imaplib
import getpass
```

```python
M = imaplib.IMAP4_SSL('imap.gmail.com')
```

```python
email = getpass.getpass('Email: ')
password = getpass.getpass('Password: ')
```

```python
M.login(email, password)
```

```python
M.list()
```

```python
M.select('inbox')
```

```python
typ, data = M.search(None, 'BEFORE 01-Nov-2000') #example 1
typ, data = M.search(None, 'FROM user@example.com') #example 2
typ, data = M.search(None, 'FROM user') #example 3

# refer to the liberary documentation for more extensive usage
```
