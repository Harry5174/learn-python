## try-except

```python
try:
    # Want to this code which may have an error
    result = 10 + 10
except:
    print("It looks like you aren't adding correctly!")
else:
    print('add went well')
    print(result)
```

```python
result
```

```python
# finally
try:
    f = open('testfile','w')
    f.write('write a test line!')
except TypeError:
    print('There was a type error!')
except OSError:
    print('Hey you have an OS error')
finally: # this block always runs regardless of an error
    print('I always run')
    
f.close()
```

```python
def ask_for_int():
    while True:
        try:
            result = int(input("Please provide a number: "))
        except:
            print("Woops! that is not a number.")
            continue
        else:
            print('Yes, Thankyou!')
            break
        finally:
            print("I will always run at the end")
```

```python
ask_for_int()
```
