```python
from random import choice

row_one = [' ', ' ', ' ']
row_two = [' ', ' ', ' ']
row_three = [' ', ' ', ' ']

# To restart the game properly if cells are rerun, ensure choices are full initially
remaining_choices = [1, 2, 3, 4, 5, 6, 7, 8, 9]

def display_table():
    print(row_one)
    print(row_two)
    print(row_three)

def display_indexs():
    print('\n ----- indexs -----')
    print([1, 2, 3])
    print([4, 5, 6])
    print([7, 8, 9])
```

```python
def check_winner():
    lines = [
        row_one, row_two, row_three, # rows
        [row_one[0], row_two[0], row_three[0]], [row_one[1], row_two[1], row_three[1]], [row_one[2], row_two[2], row_three[2]], # cols
        [row_one[0], row_two[1], row_three[2]], [row_one[2], row_two[1], row_three[0]] # diagonals
    ]
    for line in lines:
        if line[0] == line[1] == line[2] and line[0] != ' ':
            return line[0]
    return None
```

```python
def get_choice(is_user_turn):
    if is_user_turn:
        choice_str = 'string'
        while not choice_str.isdigit() or int(choice_str) not in remaining_choices:
            choice_str = input("Please enter a number from remaining choices: ")
            if not choice_str.isdigit():
                print("Sorry that is not a digit!")
            elif int(choice_str) not in remaining_choices:
                print("Sorry, that choice is not available.")
        return int(choice_str)
    else:
        # Machine choice
        return choice(remaining_choices)
```

```python
def make_move(position, symbol):
    match position:
        case 1: row_one[0] = symbol
        case 2: row_one[1] = symbol
        case 3: row_one[2] = symbol
        case 4: row_two[0] = symbol
        case 5: row_two[1] = symbol
        case 6: row_two[2] = symbol
        case 7: row_three[0] = symbol
        case 8: row_three[1] = symbol
        case 9: row_three[2] = symbol
    
    remaining_choices.remove(position)
```

```python
def main():
    is_user_turn = True  # True for user ('o'), False for machine ('x')
    
    print("Welcome to Tic Tac Toe!")
    display_indexs()
    
    while True:
        print("\nCurrent Board:")
        display_table()
        
        # Decide symbol and player name
        symbol = 'o' if is_user_turn else 'x'
        player_name = "User" if is_user_turn else "Machine"
        
        print(f"\n{player_name}'s turn ({symbol}):")
        
        # Get choice dynamically mapping same function for both
        position = get_choice(is_user_turn)
        print(f"{player_name} chose: {position}")
        
        # Make move
        make_move(position, symbol)
        
        # Check winner
        winner = check_winner()
        if winner:
            print("\nFinal Board:")
            display_table()
            if winner == 'o':
                print("User wins!")
            else:
                print("Machine wins!")
            break
            
        # Check draw
        if len(remaining_choices) == 0:
            print("\nFinal Board:")
            display_table()
            print("It's a draw!")
            break
            
        # Switch turn flag
        is_user_turn = not is_user_turn

main()
```
