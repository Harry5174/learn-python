# Map function

```python
def square(num):
    return num**2

my_nums = [1,2,3,4,5]
```

```python
map(square, my_nums)
```

```python
for item in map(square, my_nums):
    print(item)
```

```python
list(map(square, my_nums))
```

```python
def splicer(mystring):
    if len(mystring)%2 == 0:
        return 'EVEN'
    else:
        return mystring[0]
```

```python
names = ['Harry', 'Eve', 'John']

list(map(splicer, names))
```

# Filter Function

```python
def check_even(num):
    return num%2==0

my_nums = [1,2,3,4,5]
```

```python
filter(check_even, my_nums)
```

```python
list(filter(check_even, my_nums))
```

```python
for n in filter(check_even, my_nums):
    print(n)
```

# Lambda Function

```python
square = lambda num: num**2 # anonymous function so this assingment is unusual
```

```python
square(2)
```

```python
list(map(lambda num:num**2, my_nums))
```

```python
list(map(lambda name: name[0], names))
```

# Scopes 

```python
x = 50

def func():
    global x

    print(f'X is {x}')

    # Local Re-assignment with global keyword makes changes x globally!
    x = 200
    print(f'I JUST LOCALLY CHANGED X TO {x}')
```

```python
func()
```

```python
print(x)
```
