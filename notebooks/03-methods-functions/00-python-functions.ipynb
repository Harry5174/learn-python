```python
def even_check(num: int):
    return num%2 == 0

even_check(2)
```

```python
from typing import List
def check_even_list(num_list: List[int]):
    return any(x%2 == 0 for x in num_list)

check_even_list([1,2,3,4])
```

```python
# return all even numbers in a list

def return_even_list(numlist: List[int])-> List[int]:
    return [x for x in numlist if x%2 == 0]

return_even_list([1,2,3,4,5,6])
```

# tuple unpacking with functions

```python
stock_prices = [('APPL', 200), ('GOOG', 400), ('MSFT', 100)]

# for index, item in enumerate(stock_prices):
#     print(item if index == 1 else pass)

for ticker, price in stock_prices:
    print(price+(price*0.1))

def check_highest_value(stock_prices):
    current_max, highest_stock = stock_prices[0]

    for stock, price in stock_prices:
        if highest_stock < price:
            current_max, highest_stock = stock, price

    return current_max, highest_stock

check_highest_value(stock_prices)
```

# *args and **kargs 

```python
def myfunc(a, b):
    # returns 5% of the sum of a & b
    return sum((a, b)) * 0.05

myfunc(40, 60)
```

```python
def myfunc(*args): # args can be any variable name as long as followed by *
    # returns 5% of the sum of a & b
    print(args) # is a tuple
    return sum(args) * 0.05

myfunc(40, 60, 50, 100)
```

```python
def myfunc(**kwargs):
    print(kwargs)
    if 'fruit' in kwargs:
        print('My fruit of choice is {}'.format(kwargs['fruit']))
    else:
        print('no fruit found!')

myfunc(fruit='apple', veggies='lettuce')
```

```python
def show_info(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

show_info(1, 2, 3, name="Ali", age=25)
```

```python
def myfunc(*args, **kwargs):
    print(args)
    print(kwargs)
    print('I would like {} {}'.format(args[0], kwargs['food']))

myfunc(10, 20, 30, fruit='orange', food='eggs', animals='dog')
```

```python
# Define a function called myfunc that takes in an arbitrary number of arguments, and returns a list containing only those arguments that are even.

def myfunc(*args):
    return [x for x in args if x%2 ==0]
```
