# Prime Number

## Problem Statement

Given an integer `n`, return `true` if `n` is a prime number, otherwise return `false`.

A prime number is a number greater than `1` that has exactly two factors:

- `1`
- itself

---

## Examples

### Example 1

Input:
n = 7

Output:
true

Explanation:
7 has only two factors: 1 and 7.

---

### Example 2

Input:
n = 10

Output:
false

Explanation:
10 is divisible by 2 and 5.

---

### Example 3

Input:
n = 1

Output:
false

Explanation:
1 is not a prime number.

---

### Example 4

Input:
n = 2

Output:
true

Explanation:
2 is the smallest prime number.

---

## Constraints

- `-10^9 <= n <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(sqrt(n))`
- Space Complexity: `O(1)`

```python
import math

class Solution:
    def is_prime(self, n):
        if n <= 1:
            return False
        if n <= 3:
            return True
        if n%2 == 0 or n%3 == 0:
            return False
        
        limit = int(math.sqrt(n)) + 1
        
        for x in range(5, limit, 6):
            if n%x == 0 or n%(x + 2) == 0:
                return False
            
        return True
        
```

# Factorial

## Problem Statement

Given a non-negative integer `n`, return the factorial of `n`.

The factorial of a number `n` is the product of all positive integers from `1` to `n`.

It is represented as:

n! = n × (n - 1) × (n - 2) × ... × 1

Special case:

0! = 1

---

## Examples

### Example 1

Input:
n = 5

Output:
120

Explanation:
5! = 5 × 4 × 3 × 2 × 1 = 120

---

### Example 2

Input:
n = 0

Output:
1

Explanation:
0! is defined as 1.

---

### Example 3

Input:
n = 3

Output:
6

Explanation:
3! = 3 × 2 × 1 = 6

---

## Constraints

- `0 <= n <= 20`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)` for iterative solution

```python
class Solution:
    def factorial(self, n: int) -> int:
        result = 1

        for i in range(1, n + 1):
            result *= i

        return result
```

# Fibonacci Number

## Problem Statement

Given an integer `n`, return the `n`th Fibonacci number.

The Fibonacci sequence is defined as:

F(0) = 0  
F(1) = 1  

For `n > 1`:

F(n) = F(n - 1) + F(n - 2)

---

## Examples

### Example 1

Input:
n = 2

Output:
1

Explanation:
F(2) = F(1) + F(0) = 1 + 0 = 1

---

### Example 2

Input:
n = 3

Output:
2

Explanation:
F(3) = F(2) + F(1) = 1 + 1 = 2

---

### Example 3

Input:
n = 4

Output:
3

Explanation:
F(4) = F(3) + F(2) = 2 + 1 = 3

---

### Example 4

Input:
n = 0

Output:
0

---

## Constraints

- `0 <= n <= 30`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

```python
class Solution:
    def fibonacci(self, n):
        if n == 0:
            return 0
        elif n == 1:
            return 1
        
        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        
        return b
```

# Sum of Digits

## Problem Statement

Given a non-negative integer `n`, return the sum of its digits.

You should solve this using basic mathematical operations, not by converting the number into a string.

---

## Examples

### Example 1

Input:
n = 123

Output:
6

Explanation:
1 + 2 + 3 = 6

---

### Example 2

Input:
n = 405

Output:
9

Explanation:
4 + 0 + 5 = 9

---

### Example 3

Input:
n = 0

Output:
0

Explanation:
The sum of digits of 0 is 0.

---

### Example 4

Input:
n = 9999

Output:
36

Explanation:
9 + 9 + 9 + 9 = 36

---

## Constraints

- `0 <= n <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(d)`
- Space Complexity: `O(1)`

Where `d` is the number of digits in `n`.

```python
class Solution:
    def sum_digits(self, n):
        total = 0
        
        while n > 0:
            digit = n % 10 # to extract last digit
            total += digit
            
            n = n // 10 # to remove last digit
        return total
```

# Reverse Integer

## Problem Statement

Given an integer `n`, return the integer formed by reversing its digits.

You should solve this using mathematical operations, not by converting the number into a string.

For this version, assume `n` is non-negative.

---

## Examples

### Example 1

Input:
n = 123

Output:
321

Explanation:
The digits of `123` reversed become `321`.

---

### Example 2

Input:
n = 120

Output:
21

Explanation:
Reversing `120` gives `021`, which is simply `21`.

---

### Example 3

Input:
n = 0

Output:
0

---

### Example 4

Input:
n = 9070

Output:
709

---

## Constraints

- `0 <= n <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(d)`
- Space Complexity: `O(1)`

Where `d` is the number of digits in `n`.

```python
class Solution:
    def reverse_int(self, n: int) -> int:
        rev = 0
        
        while n > 0:
            digit = n % 10 # to extract last digit
            rev = rev * 10 + digit  # add in reverse order
            
            n = n // 10 # to remove last digit
        return rev
```

# Count Digits

## Problem Statement

Given a non-negative integer `n`, return the number of digits in `n`.

You should solve this using mathematical operations, not by converting the number into a string.

---

## Examples

### Example 1

Input:
n = 12345

Output:
5

Explanation:
The number `12345` has 5 digits.

---

### Example 2

Input:
n = 7

Output:
1

Explanation:
The number `7` has 1 digit.

---

### Example 3

Input:
n = 0

Output:
1

Explanation:
The number `0` has 1 digit.

---

### Example 4

Input:
n = 1000

Output:
4

---

## Constraints

- `0 <= n <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(d)`
- Space Complexity: `O(1)`

Where `d` is the number of digits in `n`.

```python
class Solution:
    def count_digits(self, n: int) -> int:
        if n == 0:
            return 1

        count = 0

        while n > 0:
            count += 1
            n //= 10

        return count
```
