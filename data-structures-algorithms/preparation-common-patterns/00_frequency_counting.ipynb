## 1. Frequency counting

Used in:
- anagram
- duplicate detection
- first non-repeating character
- most frequent element
- word count

## example

```python
def char_freq(s):
    freq = {}
    
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
        
    return freq

## OR
from collections import defaultdict

def char_freq_count(s):
    freq = defaultdict(int)
    
    for ch in s:
        freq[ch] += 1
        
    return dict(freq)
```

## Counter from collections

```python
from collections import Counter

def char_freq_counter(s):
    return dict(Counter(s))

print(char_freq_counter('banana'))
```

```python
from collections import Counter

sentence = "the quick brown fox jumps over the lazy dog the fox was quick"
words = sentence.split()
word_counts = Counter(words)

print(word_counts.most_common(3))
# Output: [('the', 3), ('quick', 2), ('fox', 2)]
```

```python
from collections import Counter

numbers = [1, 2, 2, 3, 3, 3, 4]
count_numbers = Counter(numbers)
print(count_numbers)
# Output: Counter({3: 3, 2: 2, 1: 1, 4: 1})
```

# Valid Anagram

## Problem Statement

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

An **anagram** is a word or phrase formed by rearranging the letters of another word, using all the original letters exactly once.

---

## Examples

### Example 1
Input:
s = "anagram"
t = "nagaram"

Output:
true

---

### Example 2
Input:
s = "rat"
t = "car"

Output:
false

---

## Constraints

- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters.

---

## Follow-up

What if the inputs contain Unicode characters? How would you adapt your solution?

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(n)`

```python
# Input: s = "anagram", t = "nagaram"
# Output: true 


# This soltion does not use Hashmap and is not optimal
class Solution:
    def anagram(self, s, t):
        """check if given inputs are anagram

        Args:
            :type s: str
            :type t: str
            :rtype: bool #True or False
        """
        # convert into mutable iterable
        t = list(t)

        if len(t) != len(s):
            return False

        for ch in s:
            if ch not in t:
                return False
            
            t.remove(ch)
            # print(t)

        if len(t) == 0:
            return True
```

```python
# This solution uses Sort and is 0(nlog(n))
class Solution:
    def anagram(self, s, t):
        """check if given inputs are anagram

        Args:
            :type s: str
            :type t: str
            :rtype: bool #True or False
        """
        
        return sorted(s) == sorted(t)
```

```python
# Best optimal, uses Hashmap and is O(n)
from collections import Counter
class Solution:
    def anagram(self, s, t):
        """check if given inputs are anagram

        Args:
            :type s: str
            :type t: str
            :rtype: bool #True or False
        """
        
        return Counter(s) == Counter(t)
```

```python
# non library best version

class Solution:
    def anagram(self, s, t):

        if len(s) != len(t):
            return False

        freq = {}

        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1

        for ch in t:
            if ch not in freq:
                return False

            freq[ch] -= 1

            if freq[ch] < 0:
                return False

        return True
```

# Contains Duplicate

## Problem Statement

Given an integer array `nums`, return `true` if any value appears at least twice in the array, and return `false` if every element is distinct.

---

## Examples

### Example 1
Input:
nums = [1,2,3,1]

Output:
true

Explanation:
The element `1` appears more than once.

---

### Example 2
Input:
nums = [1,2,3,4]

Output:
false

Explanation:
All elements are distinct.

---

### Example 3
Input:
nums = [1,1,1,3,3,4,3,2,4,2]

Output:
true

---

## Constraints

- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(n)`

```python
class Solution:
    def contains_dup(self, nums):
        """returns true if contains duplicate elements

        Args:
            :type nums: int
            :rtype : bool
        """
        seen = set()
        
        for num in nums:
            if nums in seen:
                return True
        
        seen.add(num)
            
        return False
    
# OR

class Solution:
    def contains_dup(self, nums):
        seen = set()
        return any(num in seen or seen.add(num) for num in nums)
    
# OR
class Solution:
    def contains_dup(self, nums):
        return len(nums) != len(set(nums))
```

```python
nums = [1,2,3,1]
sol = Solution()

sol.contains_dup(nums)
```

# First Non-Repeating Character

## Problem Statement

Given a string `s`, return the index of the first non-repeating character in it.

If every character repeats, return `-1`.

A non-repeating character is a character that appears exactly once in the string.

---

## Examples

### Example 1

Input:
s = "leetcode"

Output:
0

Explanation:
The character `"l"` appears only once and is the first non-repeating character.

---

### Example 2

Input:
s = "loveleetcode"

Output:
2

Explanation:
The character `"v"` appears only once and appears before the other non-repeating characters.

---

### Example 3

Input:
s = "aabb"

Output:
-1

Explanation:
Every character appears more than once.

---

## Constraints

- `1 <= s.length <= 10^5`
- `s` consists of lowercase English letters.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)` if only lowercase English letters  
- Space Complexity: `O(n)` for general characters

```python
class Solution:
    def first_non_repeating_character(self, s):
        freq = {}
        
        # counting frequency
        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1
            
        for i, ch in enumerate(s):
            if freq[ch] == 1:
                return i
        
        return -1
        
        
```

```python
s = 'loveleetcode'
solution = Solution()

solution.first_non_repeating_character(s)
    
```
