## example

```python
def char_freq(s):
    freq = {}
    
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
        
    return freq

## OR
from collections import defaultdict

def char_freq_count(s):
    freq = defaultdict(int)
    
    for ch in s:
        freq[ch] += 1
        
    return dict(freq)
```

## Counter from collections

```python
from collections import Counter

def char_freq_counter(s):
    return dict(Counter(s))

print(char_freq_counter('banana'))
```

## Problem Statement

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

An **anagram** is a word or phrase formed by rearranging the letters of another word, using all the original letters exactly once.

---

## Examples

### Example 1
Input:
s = "anagram"
t = "nagaram"

Output:
true

---

### Example 2
Input:
s = "rat"
t = "car"

Output:
false

---

## Constraints

- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(n)`

```python
# Input: s = "anagram", t = "nagaram"
# Output: true 


# This soltion does not use Hashmap and is not optimal
class Solution:
    def anagram(self, s, t):
        """check if given inputs are anagram

        Args:
            :type s: str
            :type t: str
            :rtype: bool #True or False
        """
        # convert into mutable iterable
        t = list(t)

        if len(t) != len(s):
            return False

        for ch in s:
            if ch not in t:
                return False
            
            t.remove(ch)
            # print(t)

        if len(t) == 0:
            return True
```

```python
# This solution uses Sort and is 0(nlog(n))
class Solution:
    def anagram(self, s, t):
        """check if given inputs are anagram

        Args:
            :type s: str
            :type t: str
            :rtype: bool #True or False
        """
        
        return sorted(s) == sorted(t)
```

# Contains Duplicate

## Problem Statement

Given an integer array `nums`, return `true` if any value appears at least twice in the array, and return `false` if every element is distinct.

---

## Examples

### Example 1
Input:
nums = [1,2,3,1]

Output:
true

Explanation:
The element `1` appears more than once.

---

### Example 2
Input:
nums = [1,2,3,4]

Output:
false

Explanation:
All elements are distinct.

---

### Example 3
Input:
nums = [1,1,1,3,3,4,3,2,4,2]

Output:
true

---

## Constraints

- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(n)`

```python
class Solution:
    def contains_dup(self, nums):
        """returns true if contains duplicate elements

        Args:
            :type nums: int
            :rtype : bool
        """
        seen = set()
        
        for num in nums:
            if nums in seen:
                return True
        
        seen.add(num)
            
        return False
    
# OR

class Solution:
    def contains_dup(self, nums):
        seen = set()
        return any(num in seen or seen.add(num) for num in nums)
    
# OR
class Solution:
    def contains_dup(self, nums):
        return len(nums) != len(set(nums))
```

```python
nums = [1,2,3,1]
sol = Solution()

sol.contains_dup(nums)
```

# First Non-Repeating Character

## Problem Statement

Given a string `s`, return the index of the first non-repeating character in it.

If every character repeats, return `-1`.

A non-repeating character is a character that appears exactly once in the string.

---

## Examples

### Example 1

Input:
s = "leetcode"

Output:
0

Explanation:
The character `"l"` appears only once and is the first non-repeating character.

---

### Example 2

Input:
s = "loveleetcode"

Output:
2

Explanation:
The character `"v"` appears only once and appears before the other non-repeating characters.

---

### Example 3

Input:
s = "aabb"

Output:
-1

Explanation:
Every character appears more than once.

---

## Constraints

- `1 <= s.length <= 10^5`
- `s` consists of lowercase English letters.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)` if only lowercase English letters  
- Space Complexity: `O(n)` for general characters

```python
class Solution:
    def firstUniqChar(self, s: str) -> int:
        freq = {}
        
        # counting frequency
        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1
            
        for i, ch in enumerate(s):
            if freq[ch] == 1:
                return i
        
        return -1
        
        
```

```python
s = 'loveleetcode'
solution = Solution()

solution.firstUniqChar(s)
```

# Most Frequent Element

## Problem Statement

Given an integer array `nums`, return the element that appears the most frequently.

If there are multiple elements with the same highest frequency, return any one of them.

---

## Examples

### Example 1

Input:
nums = [1, 2, 2, 3, 3, 3]

Output:
3

Explanation:
The number `3` appears 3 times, which is more than any other number.

---

### Example 2

Input:
nums = [4, 4, 1, 1, 2]

Output:
4

Explanation:
Both `4` and `1` appear 2 times. Since there is a tie, returning either `4` or `1` is acceptable.

---

### Example 3

Input:
nums = [7]

Output:
7

Explanation:
There is only one element, so it is the most frequent.

---

## Constraints

- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(n)`

```python
from collections import Counter

class Solution:
    def most_frequent(self, nums):
        return Counter(nums).most_common(1)[0][0]
    
## OR
class Solution:
    def most_frequent(self, nums):
        freq = {}
        max_count = 0
        most_frequent_num = None

        for num in nums:
            freq[num] = freq.get(num, 0) + 1

            if freq[num] > max_count:
                max_count = freq[num]
                most_frequent_num = num

        return most_frequent_num
```

```python
nums = [1, 2, 2, 3, 3, 3]

sol = Solution()
sol.most_frequent(nums)
```

# Word Count

## Problem Statement

Given a string `text`, return a dictionary/hashmap containing the frequency of each word.

Words are separated by spaces. The count should represent how many times each word appears in the string.

For this problem, assume:
- Words are case-sensitive.
- Punctuation is considered part of the word.
- Words are separated by one or more spaces.

---

## Examples

### Example 1

Input:
text = "hello world hello"

Output:
{
  "hello": 2,
  "world": 1
}

Explanation:
The word `"hello"` appears 2 times and `"world"` appears 1 time.

---

### Example 2

Input:
text = "python is fun python is powerful"

Output:
{
  "python": 2,
  "is": 2,
  "fun": 1,
  "powerful": 1
}

---

### Example 3

Input:
text = "one"

Output:
{
  "one": 1
}

---

### Example 4

Input:
text = ""

Output:
{}

Explanation:
There are no words in the string.

---

## Constraints

- `0 <= text.length <= 10^5`
- `text` may contain lowercase letters, uppercase letters, punctuation, and spaces.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(k)`

Where:
- `n` is the number of characters in the string
- `k` is the number of unique words

```python
class Solution:
    def word_freq(self, text):
        
        text = text.split()
        freq = {}
        
        for word in text:
            freq[word] = freq.get(word, 0) + 1
            
        return freq
```

```python
text = "hello world hello"

sol=Solution()
sol.word_freq(text)
```
