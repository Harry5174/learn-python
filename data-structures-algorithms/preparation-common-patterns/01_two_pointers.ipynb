# Valid Palindrome

## Problem Statement

Given a string `s`, return `True` if the string is a palindrome, otherwise return `False`.

A palindrome is a string that reads the same forward and backward.

For this beginner version, assume:
- The string contains only lowercase English letters.
- No spaces are included.
- No punctuation is included.

---

## Examples

### Example 1

Input:
s = "madam"

Output:
True

Explanation:
Reading `"madam"` from left to right and right to left gives the same string.

---

### Example 2

Input:
s = "racecar"

Output:
True

Explanation:
`"racecar"` is the same forward and backward.

---
## Constraints

- `0 <= s.length <= 10^5`
- `s` contains only lowercase English letters.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the string

```python
class Solution:
    def is_palindrome(self, s):
        left = 0
        right = len(s) - 1
        
        while left < right:
            if s[left] != s[right]:
                return False
            
            left += 1
            right -= 1
            
                
        return True
```

```python
s = 'madam'

sol = Solution()
sol.is_palindrome(s)
```

# Reverse String

## Problem Statement

Given a list of characters `s`, reverse the list in-place.

You must modify the original list directly and should not create a new list.

For this beginner version, assume:
- The input is a list of single characters.
- The list may be empty.
- You should return the modified list after reversing it.

---

## Examples

### Example 1

Input:
s = ["h", "e", "l", "l", "o"]

Output:
["o", "l", "l", "e", "h"]

Explanation:
The characters are reversed in-place.

---

### Example 2

Input:
s = ["H", "a", "n", "n", "a", "h"]

Output:
["h", "a", "n", "n", "a", "H"]

---

## Constraints

- `0 <= len(s) <= 10^5`
- Each element of `s` is a single character.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list

```python
from typing import List
class Solution:    
    def str_reverse(self, s: List[str]) -> List[str]:
        if not s:
            return s
        
        left = 0
        right = len(s) - 1
        
        while left < right:
            s[left], s[right] = s[right], s[left]

            left += 1
            right -= 1
            
        return s
```

```python
s = list('hello')

sol = Solution()
sol.str_reverse(s)
```

# Pair Sum in Sorted Array

## Problem Statement

Given a sorted list of integers `nums` and an integer `target`, return `True` if there are two numbers in the list whose sum equals `target`.

Return `False` if no such pair exists.

For this problem, assume:
- The list is sorted in non-decreasing order.
- You only need to return `True` or `False`.
- You cannot use the same element twice.

---

## Examples

### Example 1

Input:
nums = [1, 2, 3, 4, 6]
target = 6

Output:
True

Explanation:
The pair `2 + 4 = 6`.

---

### Example 2

Input:
nums = [1, 2, 3, 9]
target = 8

Output:
False

Explanation:
No two numbers add up to `8`.

---

### Example 3

Input:
nums = [1, 3, 4, 5, 7, 10]
target = 12

Output:
True

Explanation:
The pair `5 + 7 = 12`.

---

### Example 4

Input:
nums = [2, 2, 3, 4]
target = 4

Output:
True

Explanation:
The pair `2 + 2 = 4`. These are two different elements.

---

### Example 5

Input:
nums = []
target = 5

Output:
False

Explanation:
An empty list has no pair.

---

## Constraints

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- `nums` is sorted in non-decreasing order.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list
