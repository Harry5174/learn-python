# Valid Palindrome

## Problem Statement

Given a string `s`, return `True` if the string is a palindrome, otherwise return `False`.

A palindrome is a string that reads the same forward and backward.

For this beginner version, assume:
- The string contains only lowercase English letters.
- No spaces are included.
- No punctuation is included.

---

## Examples

### Example 1

Input:
s = "madam"

Output:
True

Explanation:
Reading `"madam"` from left to right and right to left gives the same string.

---

### Example 2

Input:
s = "racecar"

Output:
True

Explanation:
`"racecar"` is the same forward and backward.

---
## Constraints

- `0 <= s.length <= 10^5`
- `s` contains only lowercase English letters.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the string

```python
class Solution:
    def is_palindrome(self, s):
        left = 0
        right = len(s) - 1
        
        while left < right:
            if s[left] != s[right]:
                return False
            
            left += 1
            right -= 1
            
                
        return True
```

```python
s = 'madam'

sol = Solution()
sol.is_palindrome(s)
```

# Reverse String

## Problem Statement

Given a list of characters `s`, reverse the list in-place.

You must modify the original list directly and should not create a new list.

For this beginner version, assume:
- The input is a list of single characters.
- The list may be empty.
- You should return the modified list after reversing it.

---

## Examples

### Example 1

Input:
s = ["h", "e", "l", "l", "o"]

Output:
["o", "l", "l", "e", "h"]

Explanation:
The characters are reversed in-place.

---

### Example 2

Input:
s = ["H", "a", "n", "n", "a", "h"]

Output:
["h", "a", "n", "n", "a", "H"]

---

## Constraints

- `0 <= len(s) <= 10^5`
- Each element of `s` is a single character.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list

```python
from typing import List
class Solution:    
    def str_reverse(self, s: List[str]) -> List[str]:
        if not s:
            return s
        
        left = 0
        right = len(s) - 1
        
        while left < right:
            s[left], s[right] = s[right], s[left]

            left += 1
            right -= 1
            
        return s
```

```python
s = list('hello')

sol = Solution()
sol.str_reverse(s)
```

# Pair Sum in Sorted Array

## Problem Statement

Given a sorted list of integers `nums` and an integer `target`, return `True` if there are two numbers in the list whose sum equals `target`.

Return `False` if no such pair exists.

For this problem, assume:
- The list is sorted in non-decreasing order.
- You only need to return `True` or `False`.
- You cannot use the same element twice.

---

## Examples

### Example 1

Input:
nums = [1, 2, 3, 4, 6]
target = 6

Output:
True

Explanation:
The pair `2 + 4 = 6`.

---

### Example 2

Input:
nums = [1, 2, 3, 9]
target = 8

Output:
False

Explanation:
No two numbers add up to `8`.

---

### Example 3

Input:
nums = [1, 3, 4, 5, 7, 10]
target = 12

Output:
True

Explanation:
The pair `5 + 7 = 12`.

---

### Example 4

Input:
nums = [2, 2, 3, 4]
target = 4

Output:
True

Explanation:
The pair `2 + 2 = 4`. These are two different elements.

---

### Example 5

Input:
nums = []
target = 5

Output:
False

Explanation:
An empty list has no pair.

---

## Constraints

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- `nums` is sorted in non-decreasing order.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list

```python
class Solution:
    def pair_sum(self, nums, target):
        seen = set()
        
        for num in nums:
            need = target - num
            
            if need in seen:
                return True
            
            seen.add(num)
        
        return False
        
        
## OR
class Solution:
    def pair_sum(self, nums, target):
        left = 0
        right = len(nums) - 1

        while left < right:
            current_sum = nums[left] + nums[right]

            if current_sum == target:
                return True
            elif current_sum < target:
                left += 1
            else:
                right -= 1

        return False
```

```python
nums = [1, 2, 3, 4, 6]
target = 11

sol = Solution()
sol.pair_sum(nums, target)
```

# Remove Duplicates from Sorted Array

## Problem Statement

Given a sorted list of integers `nums`, remove the duplicates in-place so that each unique element appears only once.

Return the number of unique elements.

You should modify the original list so that the first `k` elements contain the unique values in their original order.

For this problem, assume:
- The list is sorted in non-decreasing order.
- You must modify the list in-place.
- You only need to return the count of unique elements.
- The values after the first `k` elements do not matter.

---

## Examples

### Example 1

Input:
nums = [1, 1, 2]

Output:
2

Modified nums:
[1, 2, _]

Explanation:
There are 2 unique elements: `1` and `2`.
The first 2 positions of the list should contain `[1, 2]`.

---

### Example 2

Input:
nums = [0, 0, 1, 1, 1, 2, 2, 3]

Output:
4

Modified nums:
[0, 1, 2, 3, _, _, _, _]

Explanation:
There are 4 unique elements: `0`, `1`, `2`, and `3`.

---

### Example 3

Input:
nums = [1, 2, 3]

Output:
3

Modified nums:
[1, 2, 3]

Explanation:
All elements are already unique.

---

### Example 4

Input:
nums = []

Output:
0

Explanation:
The list is empty, so there are no unique elements.

---

### Example 5

Input:
nums = [5, 5, 5, 5]

Output:
1

Modified nums:
[5, _, _, _]

Explanation:
Only one unique value exists.

---

## Constraints

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `nums` is sorted in non-decreasing order.

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list

```python
class Solution:
    def remove_dupl(self, nums):
        if not nums:
            return 0
        
        left = 1
        right = len(nums) - 1
        
        count = 1
        slow = 1
        
        while left <= right:
            if nums[left] != nums[left-1]:
                nums[slow] = nums[left]
                
                slow += 1
                count += 1
            left += 1

        while slow <= right:
            nums[slow] = "_"
            slow += 1
            
        return count, nums
```

```python
nums = [0, 0, 1, 1, 1, 2, 2, 3,4,4,4,5,5,6]

solution = Solution()
solution.remove_dupl(nums)
```
