# Maximum Sum Subarray of Size K

## Problem Statement

Given an integer array `nums` and an integer `k`, return the maximum sum of any contiguous subarray of size `k`.

A subarray is a continuous part of an array.

---

## Examples

### Example 1

Input:
nums = [2, 1, 5, 1, 3, 2]
k = 3

Output:
9

Explanation:
The subarray `[5, 1, 3]` has the maximum sum:

5 + 1 + 3 = 9

---

### Example 2

Input:
nums = [1, 2, 3, 4, 5]
k = 2

Output:
9

Explanation:
The subarray `[4, 5]` has the maximum sum:

4 + 5 = 9

---

### Example 3

Input:
nums = [5, 5, 5, 5]
k = 2

Output:
10

Explanation:
Every subarray of size 2 has sum 10.

---

## Constraints

- `1 <= k <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

```python
class Solution:
    def maximum_subarray_sum(self, nums, k):
        window_sum = sum(nums[:k])
        max_sum = window_sum
        
        for right in range(k, len(nums)):
            window_sum += nums[right]
            window_sum -= nums[right-k]
            
            max_sum = max(max_sum, window_sum)
        
        return max_sum
        
```

# Maximum Average Subarray I

## Problem Statement

Given an integer array `nums` consisting of `n` elements and an integer `k`, find the maximum average value of any contiguous subarray of length `k`.

Return the maximum average value.

A subarray is a continuous part of an array.

---

## Examples

### Example 1

Input:
nums = [1, 12, -5, -6, 50, 3]
k = 4

Output:
12.75

Explanation:
The subarray `[12, -5, -6, 50]` has the maximum average.

Sum:
12 + (-5) + (-6) + 50 = 51

Average:
51 / 4 = 12.75

---

### Example 2

Input:
nums = [5]
k = 1

Output:
5.0

---

### Example 3

Input:
nums = [0, 4, 0, 3, 2]
k = 1

Output:
4.0

---

## Constraints

- `1 <= k <= nums.length <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`
