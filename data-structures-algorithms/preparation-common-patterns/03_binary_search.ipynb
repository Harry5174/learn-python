# Binary Search

## Problem Statement

Given a sorted list of integers `nums` and an integer `target`, return the index of `target` if it exists in the list.

If `target` does not exist, return `-1`.

For this problem, assume:
- The list is sorted in ascending order.
- All elements are unique.
- You must solve it using binary search.

---

## Examples

### Example 1

Input:
nums = [-1, 0, 3, 5, 9, 12]
target = 9

Output:
4

Explanation:
The value `9` exists at index `4`.

---

### Example 2

Input:
nums = [-1, 0, 3, 5, 9, 12]
target = 2

Output:
-1

Explanation:
The value `2` does not exist in the list.

---

### Example 3

Input:
nums = [1]
target = 1

Output:
0

Explanation:
The value `1` exists at index `0`.

---

### Example 4

Input:
nums = [1]
target = 2

Output:
-1

Explanation:
The value `2` does not exist in the list.

---

### Example 5

Input:
nums = []
target = 5

Output:
-1

Explanation:
The list is empty, so the target cannot be found.

---

## Constraints

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- `nums` is sorted in ascending order.
- All elements in `nums` are unique.

---

## Expected Complexity

- Time Complexity: `O(log n)`
- Space Complexity: `O(1)`

Where:
- `n` is the length of the list

```python
class Solution:
    def binary_search(self, nums, target):
        left = 0
        right = len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        
        return -1
```

```python
nums = [-1, 0, 3, 5, 9, 12]
target = 2

sol = Solution()
sol.binary_search(nums, target)
```

# Search Insert Position

## Problem Statement

Given a sorted list of distinct integers `nums` and an integer `target`, return the index if the target is found.

If the target is not found, return the index where it would be inserted in order.

For this problem, assume:
- The list is sorted in ascending order.
- All elements are distinct.
- You must solve it in `O(log n)` time.

---

## Examples

### Example 1

Input:
nums = [1, 3, 5, 6]
target = 5

Output:
2

Explanation:
The target `5` is found at index `2`.

---

### Example 2

Input:
nums = [1, 3, 5, 6]
target = 2

Output:
1

Explanation:
The target `2` is not found. It should be inserted at index `1`, between `1` and `3`.

---

### Example 3

Input:
nums = [1, 3, 5, 6]
target = 7

Output:
4

Explanation:
The target `7` is greater than all elements, so it should be inserted at the end.

---

### Example 4

Input:
nums = [1, 3, 5, 6]
target = 0

Output:
0

Explanation:
The target `0` is smaller than all elements, so it should be inserted at the beginning.

---

### Example 5

Input:
nums = []
target = 4

Output:
0

Explanation:
The list is empty, so the target should be inserted at index `0`.

---

## Constraints

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- `nums` is sorted in ascending order.
- All values in `nums` are distinct.

---

## Expected Complexity

- Time Complexity: `O(log n)`
- Space Complexity: `O(1)`

```python
class Solution:
    def search_insert_pos(self, nums, target):
        left = 0
        right = len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                return mid
            
            if nums[mid] < target:
                left = mid + 1
                
            if nums[mid] > target:
                right = mid - 1
            
        return left
```
