# Second Largest Element

## Problem Statement

Given an integer array `nums`, return the second largest distinct element in the array.

If the second largest distinct element does not exist, return `-1`.

---

## Examples

### Example 1

Input:
nums = [10, 5, 20, 8]

Output:
10

Explanation:
The largest element is `20`, and the second largest distinct element is `10`.

---

### Example 2

Input:
nums = [5, 5, 5]

Output:
-1

Explanation:
There is no second largest distinct element.

---

### Example 3

Input:
nums = [2, 1]

Output:
1

Explanation:
The largest element is `2`, and the second largest is `1`.

---

### Example 4

Input:
nums = [12, 35, 1, 10, 34, 1]

Output:
34

Explanation:
The largest element is `35`, and the second largest distinct element is `34`.

---

## Constraints

- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

```python
from typing import List
class Solution:
    def second_largest(self, nums : List[int]) -> int:
        largest = float('-inf')
        second = float('-inf')
        
        for num in nums:
            if num > largest:
                second = largest
                largest = num
            
            elif largest > num > second:
                second = num
                
        return second if second != float('-inf') else -1
```

# Merge Two Sorted Arrays

## Problem Statement

Given two sorted integer arrays `nums1` and `nums2`, return a new sorted array that contains all elements from both arrays.

Both arrays are sorted in non-decreasing order.

---

## Examples

### Example 1

Input:
nums1 = [1, 3, 5]
nums2 = [2, 4, 6]

Output:
[1, 2, 3, 4, 5, 6]

---

### Example 2

Input:
nums1 = [1, 2, 3]
nums2 = [4, 5, 6]

Output:
[1, 2, 3, 4, 5, 6]

---

### Example 3

Input:
nums1 = []
nums2 = [1, 2, 3]

Output:
[1, 2, 3]

---

### Example 4

Input:
nums1 = [1, 2, 2]
nums2 = [2, 3]

Output:
[1, 2, 2, 2, 3]

---

## Constraints

- `0 <= nums1.length, nums2.length <= 10^5`
- `-10^9 <= nums1[i], nums2[i] <= 10^9`
- `nums1` and `nums2` are sorted in non-decreasing order.

---

## Expected Complexity

- Time Complexity: `O(n + m)`
- Space Complexity: `O(n + m)`

Where:
- `n = len(nums1)`
- `m = len(nums2)`

```python
class Solution:
    def merge_sorted_arr(self, nums1, nums2):
        i = 0
        j = 0
        merged = []
        
        while i < len(nums1) and j < len(nums2):
            if nums1[i] <= nums2[j]:
                merged.append(nums1[i])
                i += 1
            
            else:
                merged.append(nums2[j])
                j += 1
                
        while i < len(nums1):
            merged.append(nums1[i])
            i += 1
            
        while j < len(nums2):
            merged.append(nums2[j])
            j += 1
            
        return merged
```

# Sort 0s, 1s, and 2s

## Problem Statement

Given an array `nums` containing only `0`, `1`, and `2`, sort the array in-place so that all `0`s come first, followed by all `1`s, then all `2`s.

You must solve this without using the built-in sort function.

This is also known as the **Dutch National Flag** problem.

---

## Examples

### Example 1

Input:
nums = [2, 0, 2, 1, 1, 0]

Output:
[0, 0, 1, 1, 2, 2]

---

### Example 2

Input:
nums = [2, 0, 1]

Output:
[0, 1, 2]

---

### Example 3

Input:
nums = [0]

Output:
[0]

---

### Example 4

Input:
nums = [1, 2, 0, 1, 2, 0]

Output:
[0, 0, 1, 1, 2, 2]

---

## Constraints

- `1 <= nums.length <= 10^5`
- `nums[i]` is either `0`, `1`, or `2`

---

## Expected Complexity

- Time Complexity: `O(n)`
- Space Complexity: `O(1)`

```python
class Solution:
    def sort_colors(self, nums):
        count0 = 0
        count1 = 0
        count2 = 0

        for num in nums:
            if num == 0:
                count0 += 1
            elif num == 1:
                count1 += 1
            else:
                count2 += 1

        index = 0

        for _ in range(count0):
            nums[index] = 0
            index += 1

        for _ in range(count1):
            nums[index] = 1
            index += 1

        for _ in range(count2):
            nums[index] = 2
            index += 1

        return nums
    
## OR
class Solution:
    def sort_colors(self, nums):
        low = 0
        mid = 0
        high = len(nums) - 1

        while mid <= high:
            if nums[mid] == 0:
                nums[low], nums[mid] = nums[mid], nums[low]
                low += 1
                mid += 1

            elif nums[mid] == 1:
                mid += 1

            else:
                nums[mid], nums[high] = nums[high], nums[mid]
                high -= 1

        return nums    
```
