# Problem: Two Sum

## Link
https://leetcode.com/problems/two-sum/

## Approach
- Brute force
- Optimized (hash map)

## Complexity
Time: O(n)  
Space: O(n)

```python
nums = [2, 7, 11, 15]
target = 9
```

# Brute Force

```python
for i in range(len(nums)): # starts from 0 to len() of array. Where stop is exlusive
    for n in range(i+1, len(nums)):  # Time complexity: O(n^2)
        if nums[i]+nums[n] == target: 
            # print(nums[i],nums[n])
            print([i, n])
```

## Hash Map 

```python
seen = {}

for index, num in enumerate(nums):
    need = target - num
    
    if need in seen:
        print([seen[need], index])
    
    seen[num] = index
    
print(seen) 
```

# Leetcode

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        seen = {}
        
        for index, num in enumerate(nums):
            need = target - num
            
            if need in seen:
                return [seen[need], index]
            
            seen[num] = index
```

```python
solution = Solution()
```

```python
solution.twoSum(nums, target)
```
